```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/nlp_dataset.csv')
```

```python
df.head(5)
```

```python
df.isnull().sum()
```

```python
df['Emotion'].value_counts()
```

```python
df['cleaned_text']=df['Comment'].str.lower()
```

```python
df['cleaned_text'].head(3)
```

```python
import re
df['cleaned_text']=df['cleaned_text'].str.replace("[^a-z0-9]+"," ",regex=True)
```

```python
df['cleaned_text']=df['cleaned_text'].str.strip()
```

```python
df['cleaned_text']=df['cleaned_text'].str.replace(r'\s+'," ",regex=True)
```

```python
df['cleaned_text']=df['cleaned_text'].str.replace(r'http\S+|www\S+'," ",regex=True)
```

```python
df['cleaned_text'].head(2)
```

```python
from nltk import TweetTokenizer
tk=TweetTokenizer()
```

```python
df['tokenized_text']=df['cleaned_text'].apply(tk.tokenize)
```

```python
df['tokenized_text'].head(5)
```

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
stop_words=set(stopwords.words('english'))
```

```python
print(stop_words)
```

```python
print(len(stop_words))
```

```python
negotiation={'no','not','not'}
```

```python
stop_words=stop_words-negotiation
```

```python
print(len(stop_words))
```

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
```

```python
df['stemmed_tokens']=df['tokenized_text'].apply(lambda x: [stemmer.stem(word) for word in x if word not in stop_words])
```

```python
df['stemmed_tokens'].head(5)
```

```python
df['filtered_text']=df['stemmed_tokens'].apply(lambda x: " ".join(x))
```

```python
df['filtered_text'].head(5)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Emotion']=le.fit_transform(df['Emotion'])
```

```python
df['Emotion'].value_counts()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf=TfidfVectorizer()
vectorized=tf.fit_transform(df['filtered_text'])
```

```python
print(vectorized)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(vectorized,df['Emotion'],test_size=0.2,random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB
nb=MultinomialNB()
nb.fit(x_train,y_train)
```

```python
nave_pred=nb.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
print(accuracy_score(y_test,nave_pred))
print(classification_report(y_test,nave_pred))
```

```python
from sklearn.svm import SVC
svm=SVC()
svm.fit(x_train,y_train)
```

```python
svc_pred=svm.predict(x_test)
```

```python
print(accuracy_score(y_test,svc_pred))
print(classification_report(y_test,svc_pred))
```
